**About the data**

This dataset includes information on [Water Conflict and Crisis Events in the Colorado River Basin](https://www.sciencebase.gov/catalog/item/63acac09d34e92aad3ca1480) and was sourced from the USGS digital repository [ScienceBase](https://www.sciencebase.gov/catalog/). These data were collected from 2005-01-01 to 2021-12-31, with the purpose of understanding vulnerability to water insecurity, resiliency demonstrated by institutions, and conflict or crisis around water resource management (Holloman et al., 2023).

**Resources**

Holloman, D.V., Hines, M.K., and Zoanni, D.K., 2023, Coded Water Conflict and Crisis Events in the Colorado River Basin, Derived from LexisNexis search 2005-2021: U.S. Geological Survey data release, [https://doi.org/10.5066/P9X6WR7J]().

### String accessor for pandas.Series

```python
# Example series
s = pd.Series(['California; Nevada', 'Arizona', np.nan, 'Nevada; Utah'])
s
```

```python
# Use str accessor with additional methods to perform string operations

s.str.split(';') # Created list of of individual strings distinguished by where ; used to be, but still in same row
```

```python
s.str.split(';').explode() # Seperates individual strings in list. Preserves index, so there are duplicates
```
