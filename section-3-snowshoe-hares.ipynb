## Detecting messy values

Intended sex column values:

| Value   | Meaning             |
| ------- | --------------------|
| f       | Female              |
| m       | Male                |
| m?      | Not confirmed male  |


```python
# Number of non null sex values
hares['sex'].info()
```

```python
# Number of actual each unique value in the sex column
hares['sex'].value_counts(dropna = False) # Keep the number of NAs in the column with dropna = False
```

## Clean values

```python
# Use replace() to replace messy values with cleaner values
hares['sex_simple'] = (hares['sex']
 .replace(to_replace=['F','f','f '],value='female')
 .replace(to_replace=['M','m','m '],value='male')
 .replace(to_replace=['?', 'F?', 'f?', 'M?', 'm?', 'pf'],value=np.nan))
```

```python
hares['sex_simple'].value_counts(dropna=False)
```

## Calculate mean weight

```python
hares.groupby('sex_simple')['weight'].mean()
```

## Streamlined workflow

```python
# Import packages
import pandas as pd
import numpy as np

# Read in the data
hares = pd.read_csv('data/55_Hare_Data_2012.txt')

# Replace messy sex values with cleaner values
hares['sex_simple'] = (hares['sex']
 .replace(to_replace=['F','f','f '],value='female')
 .replace(to_replace=['M','m','m '],value='male')
 .replace(to_replace=['?', 'F?', 'f?', 'M?', 'm?', 'pf'],value=np.nan))

# Calculate mean weight for females and males
mean_sex = hares.groupby('sex_simple')['weight'].mean()
mean_sex
```
